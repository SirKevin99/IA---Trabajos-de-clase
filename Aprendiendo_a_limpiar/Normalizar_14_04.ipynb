```python
df.fillna(0)
```

```python
import unicodedata

# 1. Función para quitar tildes y caracteres especiales
def quitar_tildes(texto):
    if isinstance(texto, str):
        # Normaliza a formato NFD (separa la letra del acento) y filtra los acentos
        texto = unicodedata.normalize('NFD', texto)
        texto = texto.encode('ascii', 'ignore').decode("utf-8")
    return texto

# 2. Aplicamos la limpieza
# Primero quitamos tildes, luego pasamos a minúsculas y finalmente aplicamos Title Case
df['nombre'] = df['nombre'].apply(quitar_tildes).str.lower().str.title()

print("Columna 'nombre' normalizada:")
print(df['nombre'].head())
```

```python
# 1. Diccionario de mapeo para palabras a números
mapeo_numeros = {
    'veinte': 20,
    'veintiuno': 21,
    'veintidos': 22,
    'veintitres': 23,
    'veinticuatro': 24
}

# 2. Convertir palabras a números (si el valor existe en el mapeo)
df['edad'] = df['edad'].apply(lambda x: mapeo_numeros.get(str(x).lower(), x))

# 3. Convertir todo a numérico (esto maneja los strings como '19')
df['edad'] = pd.to_numeric(df['edad'])

# 4. Validar rango (17-30)
# Filtramos el dataframe para ver si alguien está fuera de rango o usamos una máscara
fuera_de_rango = df[(df['edad'] < 17) | (df['edad'] > 30)]

print(f"Valores fuera de rango: {len(fuera_de_rango)}")
print("\nColumna edad procesada:")
print(df[['nombre', 'edad']].head())
```

```python
# Reutilizamos la función quitar_tildes definida en el ejercicio 1
# (Asegúrate de haber corrido esa celda antes)

# 1. Quitar tildes
df['ciudad'] = df['ciudad'].apply(quitar_tildes)

# 2. Convertir a minúsculas y aplicar Title Case (Primera letra Mayúscula)
df['ciudad'] = df['ciudad'].str.strip().str.lower().str.title()

# 3. Verificamos los valores únicos para asegurar la unificación
print("Ciudades unificadas:")
print(df['ciudad'].unique())

print("\nDataset actualizado:")
print(df[['nombre', 'ciudad']].head(10))
```

```python
# Reutilizamos la función quitar_tildes definida en el ejercicio 1
# (Asegúrate de haber corrido esa celda antes)

# 1. Quitar tildes
df['carrera'] = df['carrera'].apply(quitar_tildes)

# 2. Convertir a minúsculas y aplicar Title Case (Primera letra Mayúscula)
df['carrera'] = df['carrera'].str.strip().str.lower().str.title()

# 3. Verificamos los valores únicos para asegurar la unificación
print("Carreras unificadas:")
print(df['carrera'].unique())

print("\nDataset actualizado:")
print(df[['nombre', 'ciudad' , 'carrera']].head(10))
```

```python
# 1. Convertir a float
# errors='coerce' transformaría cualquier valor no numérico en NaN (Not a Number)
df['promedio'] = pd.to_numeric(df['promedio'], errors='coerce')

# 2. Validar rango (0.0 a 10.0)
# Creamos una máscara booleana para encontrar valores fuera de rango
fuera_de_rango_prom = df[(df['promedio'] < 0) | (df['promedio'] > 10)]

# 3. Mostrar resultados
print(f"Valores fuera de rango en promedio: {len(fuera_de_rango_prom)}")
print("\nTipos de datos actualizados:")
print(df[['nombre', 'promedio']].dtypes)

print("\nPrimeros 5 promedios ya como float:")
print(df[['nombre', 'promedio']].head())
```

```python
import numpy as np

# 1. Todo en minúsculas
df['email'] = df['email'].str.lower().str.strip()

# 2. Función para validar el formato del email
def validar_email(email):
    if pd.isna(email):
        return np.nan

    # Contamos cuántas "@" tiene el texto
    conteo_arroba = email.count('@')

    # Un email es válido si tiene exactamente una "@"
    # (Podrías agregar más validaciones como que tenga un punto después, pero esto cumple tu consigna)
    if conteo_arroba == 1:
        return email
    else:
        return np.nan

# 3. Aplicar la validación y marcar inválidos como NaN
df['email'] = df['email'].apply(validar_email)

# 4. Mostrar resultados para verificar
print("Conteo de valores nulos (emails inválidos):", df['email'].isna().sum())
print("\nDataset con emails normalizados:")
print(df[['nombre', 'email']])
```

```python
# 1. Creamos un diccionario de mapeo
# Normalizamos visualmente a minúsculas para que el mapeo sea más sencillo
mapeo_booleano = {
    'si': True,
    'no': False
}

# 2. Aplicamos la transformación
# Convertimos a minúsculas, quitamos espacios y mapeamos
df['activo'] = df['activo'].str.lower().str.strip().map(mapeo_booleano)

# 3. Verificamos el resultado y el tipo de dato
print("Columna 'activo' convertida a booleano:")
print(df[['nombre', 'activo']].head())
print(f"\nTipo de dato de la columna: {df['activo'].dtype}")
```
