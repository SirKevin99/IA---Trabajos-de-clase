```python
def limpiar_texto_filas(df, columna):
    if columna in df.columns:
        # Pasar a minúsculas, quitar espacios y acentos
        df[columna] = df[columna].astype(str).str.lower().str.strip()
        df[columna] = df[columna].apply(lambda x:
            "".join(c for c in unicodedata.normalize('NFD', x)
            if unicodedata.category(c) != 'Mn'))
    return df

# Ejemplo de uso:
# df_red = limpiar_texto_filas(df_red, 'nombre_columna_texto')
```
