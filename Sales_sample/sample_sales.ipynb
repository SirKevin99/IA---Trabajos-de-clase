```python
from langchain_community.document_loaders import DataFrameLoader
from langchain_mistralai import MistralAIEmbeddings, ChatMistralAI
from langchain_community.vectorstores import Chroma
from langchain_core.tools import create_retriever_tool
from langchain_core.messages import HumanMessage, ToolMessage, SystemMessage

# ========================================================
# 1. CREAR EL CORPUS VECTORIAL (EL RAG)
# ========================================================
print("⏳ Generando los vectores del Corpus en la memoria...")
loader = DataFrameLoader(df, page_content_column="TEXT_CONTEXT_ES")
documentos = loader.load()

# Inicializamos los embeddings oficiales de Mistral
embeddings = MistralAIEmbeddings(model="mistral-embed")
# Creamos la base de datos vectorial asignándola a la variable vector_store
vector_store = Chroma.from_documents(documentos, embeddings)
print("✅ Base de datos vectorial 'vector_store' creada con éxito.")


# ========================================================
# 2. CONFIGURAR LA HERRAMIENTA Y EL MODELO MISTRAL
# ========================================================
# Ahora que existe vector_store, creamos el retriever sin perder la variable
retriever = vector_store.as_retriever(search_kwargs={"k": 3})
tool_rag = create_retriever_tool(
    retriever,
    "buscar_en_ventas",
    "Úsala para buscar información detallada sobre las ventas, clientes, productos, pedidos, países y estados de cuenta."
)

# Inicializamos Mistral y le enlazamos la herramienta
llm = ChatMistralAI(model="mistral-small-latest", temperature=0)
llm_con_herramientas = llm.bind_tools([tool_rag])
print("✅ Herramientas vinculadas a Mistral.")


# ========================================================
# 3. EL ORQUESTADOR DEL AGENTE MODERNO
# ========================================================
def ejecutar_agente_langchain(pregunta_usuario: str):
    print(f"\n💬 **Pregunta del Usuario:** {pregunta_usuario}\n")

    messages = [
        SystemMessage(content="Eres un asistente experto en análisis de datos de ventas de la empresa. "
                             "Responde siempre en español de forma clara, profesional y concisa basándote en la herramienta 'buscar_en_ventas'."),
        HumanMessage(content=pregunta_usuario)
    ]

    # Evaluar la pregunta con Mistral
    respuesta_modelo = llm_con_herramientas.invoke(messages)

    # Si el agente decide usar el RAG
    if respuesta_modelo.tool_calls:
        print("🤖 [Agente]: Necesito buscar datos en el Corpus. Activando herramienta...")
        messages.append(respuesta_modelo)

        for tool_call in respuesta_modelo.tool_calls:
            print(f"🔍 [RAG]: Recuperando registros para '{tool_call['args']['query']}'...")
            resultado_txt = tool_rag.invoke(tool_call['args'])
            messages.append(ToolMessage(content=resultado_txt, tool_call_id=tool_call['id']))

        print("✍️ [Mistral]: Redactando respuesta final basada en el CSV...")
        respuesta_final = llm_con_herramientas.invoke(messages)
        return respuesta_final.content
    else:
        return respuesta_modelo.content

print("\n🚀 ¡TODO LISTO! Estructura LangChain acoplada y lista para interrogar.")
```
